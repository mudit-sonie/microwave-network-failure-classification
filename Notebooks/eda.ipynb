```python
TARGET = "label"

X = df.drop(columns=[TARGET])
y = df[TARGET]

print(y.value_counts().sort_index())
print("\nClass proportions:")
print(y.value_counts(normalize=True).sort_index())
```

# VISUALISATION

```python
%matplotlib inline

y.value_counts().sort_index().plot(
    kind="bar",
    figsize=(7, 4)
)

plt.title("Distribution of Microwave Hardware Failures")
plt.xlabel("Failure class")
plt.ylabel("Number of observations")
plt.tight_layout()
plt.show()
```

## Checking for values outside the given range in features 

```python
print(X.describe().T)
```

```python
invalid = (X < 0) | (X > 900)

print("Values outside 0–900:", invalid.sum().sum())
```

## Inactive alarm features in most observations.

```python
zero_percentage = (X == 0).mean() * 100

print(
    zero_percentage
    .sort_values(ascending=False)
    .head(20)
)
```

```python
constant_features = X.columns[X.nunique() <= 1]

print("Constant features:", len(constant_features))
print(list(constant_features))
```

```python
# A few features with nonzero variation
%matplotlib inline

candidate_features = (
    X.nunique()
    .loc[lambda s: s > 1]
    .index[:4]
)

for feature in candidate_features:
    df.boxplot(column=feature, by=TARGET, figsize=(7, 4))
    plt.title(f"{feature}: distribution by failure class")
    plt.suptitle("")
    plt.xlabel("Failure class")
    plt.ylabel("Alarm duration (seconds)")
    plt.tight_layout()
    plt.show()
```
